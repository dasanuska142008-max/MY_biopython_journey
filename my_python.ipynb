```python
#Create a function that takes the name of a FASTA file as input and returns its content (sequences) as list
from Bio import SeqIO
sequences=[]
for seq_record in SeqIO.parse("/sequence.fasta","fasta"):
  sequences.append(seq_record)
  print(sequences)
```

```python
#Create a function that takes the name of a FASTA file as input and returns a list of FASTA identifiers
from Bio import SeqIO
fasta_id =[]
for seq_record in SeqIO.parse("/sequence.fasta","fasta"):
  fasta_id.append(seq_record.id)
  print(fasta_id)
```
